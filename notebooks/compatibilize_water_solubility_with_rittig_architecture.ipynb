```python
import pandas as pd
import numpy as np


###Load train and test set
train_set = pd.read_csv(r'..\data\solubility_water\unique_train_water_final.csv')  
test_set = pd.read_csv(r'..\data\solubility_water\unique_test_water_final.csv')

solute_list = pd.concat((train_set, test_set)).reset_index(drop=True)
solute_list
```

```python
solute_id = ['solute_'+str(i) for i in range(len(solute_list))]
solute_id
```

```python
solute_list['solute_id'] = solute_id
solute_list
```

```python
solute_list = solute_list.drop(["LogS","occurence"], axis=1)
solute_list
```

```python
solute_list = solute_list[['InChIKey','solute_id','smiles_canon']]
solute_list
```

```python
solute_list.to_csv('../data/solubility_water/solute_list.csv', index=False)
```

```python
solvent_list = pd.read_csv(r'..\data\solvent_list.csv')
water = solvent_list.loc[solvent_list['solvent_name']=='WATER'].reset_index(drop=True)
water
```

```python
water.to_csv('../data/solubility_water/solvent_list_water.csv', index=False)
```

```python
water_id = water['solvent_id'].iloc[0]
water_id
```

```python
train_set['solvent_id'] = [water_id]*len(train_set)
train_set
```

```python
train_solute_id = []
for solute in train_set['smiles_canon'].tolist():
    searched_id = solute_list[solute_list['smiles_canon'] == solute]['solute_id'].tolist()
    if len(searched_id) != 1:
        print('WARNING: conflicting id')
        break
    searched_id = searched_id[0]
    train_solute_id.append(searched_id)
train_set['solute_id'] = train_solute_id
train_set
```

```python
test_set['solvent_id'] = [water_id]*len(test_set)
test_solute_id = []
for solute in test_set['smiles_canon'].tolist():
    searched_id = solute_list[solute_list['smiles_canon'] == solute]['solute_id'].tolist()
    if len(searched_id) != 1:
        print('WARNING: conflicting id')
        break
    searched_id = searched_id[0]
    test_solute_id.append(searched_id)
test_set['solute_id'] = test_solute_id
test_set
```

```python
#remove conflicting smiles (RDKit cannot process them into mol object due to a Nitrogen atom having more valence than expected)
conflicting_smiles = ["CC1=CC=C[NH++]([O-])[CH-]1", "OC(=O)C1=C[NH++]([O-])[CH-]C=C1"]

indx = train_set[train_set["smiles_canon"].apply(lambda x : x in conflicting_smiles)]   
train_set = train_set.drop(index=indx.index)
train_set
```

```python
train_set.reset_index(drop=True)
train_set
```

```python
train_set.to_csv('../data/solubility_water/COMPATIBLE_unique_train_water_final.csv', index=False)
test_set.to_csv('../data/solubility_water/COMPATIBLE_unique_test_water_final.csv', index=False)
```
